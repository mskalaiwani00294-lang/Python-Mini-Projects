You are given a list of marks. Print the index and the square of each mark using enumerate() and list comprehension.

```python
marks = [10, 20, 30, 40]
# Your code here



# First create squares using list comprehension
squares=[mark**2 for mark in marks]

# Then enumerate over them
for index,values in enumerate(squares):

  print(index,values)
```

```python
# Create a list of squares of even numbers between 1 and 20 using list comprehension
squares = [num ** 2 for num in range(1, 21) if num % 2 == 0]

print(squares)
```
