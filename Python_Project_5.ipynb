Title: Scholarship Eligibility Checker

Instructions:

Create variables: studentname, marks, attendance, familyincome.

Use conditional statements (if, elif, else) to:

Award Full Scholarship if marks ≥ 90 and attendance ≥ 90.

Award Half Scholarship if marks ≥ 80 or attendance ≥ 85.

Award Need-Based Scholarship if marks ≥ 70 and family_income < 30000.

Otherwise, display Not Eligible.

Display all data.





# **Scholarship Eligibility Checker**

```python
#Create variables: studentname, marks, attendance, familyincome
studentname=input("Enter Student_Name")
mark=int(input("Enter mark"))
attendance=int(input("Enter attendance"))
family_income=float(input("Enter family_income"))


#Use conditional statements (if, elif, else)
#Full Scholarship if marks ≥ 90 and attendance ≥ 90.

if mark >= 90 and attendance >= 90:
    scholarship = "Full Scholarship"
 #Half Scholarship if marks ≥ 80 or attendance ≥ 85
elif mark >= 80 or attendance >= 85:
    scholarship = "Half Scholarship"
#Need-Based Scholarship if marks ≥ 70 and family_income < 30000
elif mark >= 70 and family_income < 30000:
    scholarship = "Need-Based Scholarship"
#display Not Eligible
else:
    scholarship = "Not Eligible"


print("\n Scholarship Eligibility")
print("Student Name: ",studentname)
print("Mark: ",mark)
print("Attendance: ",attendance)
print("Family_Income: ",family_income)
print("Scholarship:", scholarship)
```
