Built-in Functions and User-Defined Functions

Project Title: Simple Calculator App

Scenario:
 A company wants to provide users with a simple calculator to perform basic mathematical operations.

Tasks:

Create functions for addition, subtraction, multiplication, and division.


Ask the user to input two numbers and the operation they want to perform.


Call the appropriate function and display the result.


Use built-in functions like abs(), round(), and pow() within your

solution.
Handle division by zero gracefully and inform the user.

```python
# Addition function
def add(a,b):
  return a+b


# Subtraction function
def subtract(a,b):
  return a-b


# Multiplication function
def multiple(a,b):
  return a*b



# Division function
def devision(a,b):
  if b==0:
    return"cannot divide by zero"
    return a/b






# Get input from user
num1=float(input("Enter the 1st number:"))
num2=float(input("Enter the 2nd number:"))
choose_function=input("Enter the function(+,-,/):")

#call Appropriate fountion
if choose_function=="+":
  result=add(num1,num2)

elif choose_function=="-":
  result=subtract(num1,num2)

elif choose_function=="*":
  result=multiple(num1,num2)

elif choose_function=="/":
  result=devision(num1,num2)

else:
  result="invalid function"




# Display result
print("Result:", result)


# Built-in functions
print("Absolute value:", abs(result))
print("Rounded value:", round(result, 2))
print("Power:", pow(num1, 2))
```
