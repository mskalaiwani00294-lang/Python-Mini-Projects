Data Cleaning: Extract and Format Employee Details

Scenario: You received raw HR data containing inconsistent formatting.
You need to extract the employee name, email, and ID properly.

Data:


record = " EmployeeName :  john doe "

Email = "JOHN.DOE@company.COM"

ID =  " emp_0087 "

Task Instructions:

Use string methods and slicing to:

Remove unwanted spaces.

Convert:

Name → John Doe

Email → john.doe@company.com

Extract the Employee ID and print only the numeric part: 0087

```python
#**Data Cleaning: Extract and Format Employee Details**
record = " EmployeeName :  john doe "
Email = "JOHN.DOE@company.COM"
ID = " emp_0087 "
#Exract and clean name
name=record.split(':')[1].strip().title()
#clean Email
Email=Email.strip().lower()
#Exract numberic par of id
employee_id=ID.strip()[4:]
print('Name:',name)
print('Email:',Email)
print('employee_id:',employee_id)
```
