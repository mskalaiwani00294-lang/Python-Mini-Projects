Topic: Departmental Data Tracker

Instructions:

Create a dictionary department with keys: "dept_name", "faculty", "courses", "budget".

Add and modify items using dictionary methods (update, pop, get, etc.).

Create two sets: online_courses and offline_courses.

Perform union, intersection, difference, and symmetric_difference.

Create a frozenset named permanent_faculty.

Display all results without loops

# **Departmental Data Tracker**

```python
# Create dictionary
department = {
    "dept_name": "Data Analytics",
    "faculty": ["Anu", "Ravi", "Priya"],
    "courses": ["Python", "SQL", "Power BI"],
    "budget": 500000
}

print("Original Department:", department)
```

```python
# Add a new item using update()
department.update({"location": "Chennai"})
print("After update:", department)

# Modify an item
department.update({"budget": 600000})
print("Modified Budget:", department.get("budget"))

# Get a value
print("Department Name:", department.get("dept_name"))

# Remove an item using pop()
removed_item = department.pop("location")
print("Removed Item:", removed_item)

print("Final Department:", department)
```

```python
# Create two sets
online_courses = {"Python", "SQL", "Power BI", "Excel"}
offline_courses = {"SQL", "Excel", "Tableau", "Java"}
```

```python
 #Union
print("Union:", online_courses.union(offline_courses))

# Intersection
print("Intersection:", online_courses.intersection(offline_courses))

# Difference
print("Online Only:", online_courses.difference(offline_courses))

# Symmetric Difference
print("Different Courses:", online_courses.symmetric_difference(offline_courses))
```

```python
# Create frozenset
permanent_faculty = frozenset({"Anu", "Ravi", "Priya"})

print("Permanent Faculty:", permanent_faculty)
```
